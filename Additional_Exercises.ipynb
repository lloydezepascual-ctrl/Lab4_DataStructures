```python
# Exercise 1: Engineering Test Data Organizer
LAST_NAME = "PASCUAL"
SEED_NUM = 4321
FAVORITE_ARTIST = "ARTHUR NERY"

test_sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

artist_clean = FAVORITE_ARTIST.replace(" ", "")
values = [(ord(char) * SEED_NUM) % 100 + 10 for char in artist_clean]

raw_records = [
    (test_sources[0], categories[0], values[0]),
    (test_sources[1], categories[1], values[1]),
    (test_sources[2], categories[2], values[2]),
    (test_sources[3], categories[3], values[3]),
    (test_sources[4], categories[4], values[4]),
    (test_sources[0], categories[0], values[5]),  # Repeated record
    (test_sources[0], categories[1], values[6]),
    (test_sources[1], categories[0], values[7]),
    (test_sources[2], categories[1], values[8]),
]

print("--- 1. Generated Raw Test Data ---")
for record in raw_records:
    print(record)

organized_data = {}
for source, cat, val in raw_records:
    if source not in organized_data:
        organized_data[source] = {}
    if cat not in organized_data[source]:
        organized_data[source][cat] = []
    organized_data[source][cat].append(val)

seen_pairs = set()
repeated_pairs = set()
for source, cat, _ in raw_records:
    pair = (source, cat)
    if pair in seen_pairs:
        repeated_pairs.add(pair)
    else:
        seen_pairs.add(pair)

print("\n--- 3. Repeated & Distinct Pairs ---")
print("Distinct Pairs:", seen_pairs)
print("Repeated Pairs:", repeated_pairs)

category_summaries = {}
for cat in categories:
    cat_vals = [val for _, c, val in raw_records if c == cat]
    if cat_vals:
        category_summaries[cat] = {
            "Count": len(cat_vals),
            "Min": min(cat_vals),
            "Max": max(cat_vals),
            "Average": round(sum(cat_vals) / len(cat_vals), 2)
        }

print("\n" + "="*50)
print(" FINAL ORGANIZED AND SUMMARIZED TEST DATASET ")
print("="*50)
for source, cats in sorted(organized_data.items()):
    print(f"\nSource: {source}")
    for cat, vals in cats.items():
        avg_val = round(sum(vals)/len(vals), 2)
        print(f"  [{cat}] Values: {vals} | Avg: {avg_val}")

print("\n--- Category Overall Summaries ---")
for cat, stats in category_summaries.items():
    print(f"{cat:12} -> {stats}")
```

```python
# Exercise 2: Student Performance Data Analyzer
LAST_NAME = "PASCUAL"
SEED_NUM = 4321
FAVORITE_ARTIST = "ARTHUR NERY"

student_names = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

def get_performance_category(score):
    if score is None:
        return "INCOMPLETE"
    elif score >= 90:
        return "EXCELLENT"
    elif score >= 75:
        return "PASSING"
    else:
        return "NEEDS IMPROVEMENT"

artist_clean = FAVORITE_ARTIST.replace(" ", "")
scores = [(ord(c) * SEED_NUM) % 51 + 50 for c in artist_clean]

records = [
    (student_names[0], subjects[0], scores[0]),
    (student_names[1], subjects[1], scores[1]),
    (student_names[2], subjects[2], scores[2]),
    (student_names[3], subjects[3], scores[3]),
    (student_names[4], subjects[4], scores[4]),
    (student_names[0], subjects[0], scores[5]),  # Repeated record
    (student_names[1], subjects[2], None),        # Incomplete record
    (student_names[2], subjects[0], scores[6]),
]

processed_records = []
seen_pairs = set()
repeated_records = []
incomplete_records = []

for student, subject, score in records:
    pair = (student, subject)
    if pair in seen_pairs:
        repeated_records.append((student, subject, score))
    else:
        seen_pairs.add(pair)
        
    if score is None:
        incomplete_records.append((student, subject))
    
    category = get_performance_category(score)
    processed_records.append({
        "Student": student,
        "Subject": subject,
        "Score": score,
        "Category": category
    })

subject_stats = {}
for sub in subjects:
    sub_scores = [r["Score"] for r in processed_records if r["Subject"] == sub and r["Score"] is not None]
    if sub_scores:
        subject_stats[sub] = {
            "Average": round(sum(sub_scores) / len(sub_scores), 2),
            "Highest": max(sub_scores),
            "Lowest": min(sub_scores)
        }

print("\n" + "="*50)
print(" FINAL STUDENT PERFORMANCE REPORT ")
print("="*50)
print(f"{'Student':<12} | {'Subject':<12} | {'Score':<6} | {'Category'}")
print("-" * 50)
for r in sorted(processed_records, key=lambda x: (x["Student"], x["Subject"])):
    score_str = str(r['Score']) if r['Score'] is not None else "N/A"
    print(f"{r['Student']:<12} | {r['Subject']:<12} | {score_str:<6} | {r['Category']}")

print("\n--- Subject Level Performance Summaries ---")
for sub, stats in subject_stats.items():
    print(f"{sub:<12} -> Avg: {stats['Average']}, High: {stats['Highest']}, Low: {stats['Lowest']}")
```

```python
# Exercise 3: Transaction History Analyzer
LAST_NAME = "PASCUAL"
SEED_NUM = 4321
FAVORITE_ARTIST = "ARTHUR NERY"

customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]

product_prices = {
    "NOTEBOOK": 50.0,
    "PEN": 15.0,
    "USB": 350.0,
    "HEADPHONES": 750.0,
    "CALCULATOR": 450.0
}

artist_clean = FAVORITE_ARTIST.replace(" ", "")
quantities = [(ord(c) * SEED_NUM) % 5 + 1 for c in artist_clean]

transactions = [
    (customers[0], products[0], quantities[0]),
    (customers[1], products[1], quantities[1]),
    (customers[2], products[2], quantities[2]),
    (customers[3], products[3], quantities[3]),
    (customers[4], products[4], quantities[4]),
    (customers[0], products[0], quantities[5]),  # Repeated purchase
    (customers[1], products[3], quantities[6]),
    (customers[2], products[4], quantities[7]),
]

detailed_transactions = []
customer_totals = {}
product_totals = {}

for cust, prod, qty in transactions:
    unit_price = product_prices[prod]
    total_val = unit_price * qty
    detailed_transactions.append({
        "Customer": cust,
        "Product": prod,
        "Quantity": qty,
        "UnitPrice": unit_price,
        "TotalValue": total_val
    })
    customer_totals[cust] = customer_totals.get(cust, 0) + total_val
    product_totals[prod] = product_totals.get(prod, 0) + total_val

overall_total = sum(t["TotalValue"] for t in detailed_transactions)

print("\n" + "="*60)
print(" FINAL CONSOLIDATED TRANSACTION REPORT ")
print("="*60)
print(f"{'Customer':<12} | {'Product':<12} | {'Qty':<4} | {'Price':<8} | {'Total Value'}")
print("-" * 60)
for t in sorted(detailed_transactions, key=lambda x: x["Customer"]):
    print(f"{t['Customer']:<12} | {t['Product']:<12} | {t['Quantity']:<4} | ₱{t['UnitPrice']:<7.2f} | ₱{t['TotalValue']:<8.2f}")

print("\n--- Summary by Customer ---")
for cust, total in sorted(customer_totals.items()):
    print(f"{cust:<12} -> Total Spent: ₱{total:,.2f}")

print("\n--- Summary by Product ---")
for prod, total in sorted(product_totals.items()):
    print(f"{prod:<12} -> Total Sales: ₱{total:,.2f}")

print("-" * 60)
print(f"OVERALL TRANSACTION DATASET TOTAL: ₱{overall_total:,.2f}")
```
